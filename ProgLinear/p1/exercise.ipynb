```python
"""
Programação Linear — Cia. de Seguros Primo
===========================================

Variáveis de decisão:
    x1 = unidades de seguro de risco especial
    x2 = unidades de hipotecas

Função objetivo:
    Max Z = 5·x1 + 2·x2

Restrições (horas disponíveis por departamento):
    R1 — Subscrição     : 3x1 + 2x2 ≤ 2.400
    R2 — Administração  :       x2  ≤   800
    R3 — Ped. Indeniz.  : 2x1       ≤ 1.200  →  x1 ≤ 600
    R4 — não-negatividade: x1 ≥ 0,  x2 ≥ 0
"""
```

```python


import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import linprog
from itertools import combinations

print("=" * 56)
print("  CIA. DE SEGUROS PRIMO — Programação Linear")
print("=" * 56)

print("""
Sistema de equações (inequações):

  Função objetivo:
    Max Z = 5x₁ + 2x₂

  Sujeito a:
    R1 (Subscrição)    : 3x₁ + 2x₂ ≤ 2.400
    R2 (Administração) :        x₂  ≤   800
    R3 (Indenização)   : 2x₁        ≤ 1.200
    R4 (não-neg.)      : x₁ ≥ 0, x₂ ≥ 0
""")
```

```python
# -------------------------------------------------------
# 1. SCIPY — linprog (minimiza, então negamos os custos)
# -------------------------------------------------------
res = linprog(
    c      = [-5, -2],                        # negado para maximizar
    A_ub   = [[ 3,  2],                       # R1
              [ 0,  1],                       # R2
              [ 2,  0]],                      # R3
    b_ub   = [2400, 800, 1200],
    bounds = [(0, None), (0, None)],
    method = 'highs'
)

res_ex4 = linprog(
    c      = [-3, -2],                        # negado para maximizar
    A_ub   = [[ 3,  2],                       # R1
              [ 0,  1],                       # R2
              [ 2,  0]],                      # R3
    b_ub   = [2400, 800, 1200],
    bounds = [(0, None), (0, None)],
    method = 'highs'
)
```

```python

x1_opt, x2_opt = res.x
Z_opt = -res.fun
```

```python
print("--- Solução ---")
print(f"\n  x₁ (risco especial) = {x1_opt:.4 f} unidades")
print(f"  x₂ (hipotecas)      = {x2_opt:.4f} unidades")
print(f"\n  Lucro total máximo  : Z = $ {Z_opt:,.2f}")
```
