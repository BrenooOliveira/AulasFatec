```python
from dataclasses import dataclass


@dataclass(frozen=True)
class GameState:
    board: tuple[str, ...]
    player: str

    def possible_moves(self):
        return tuple(
            i
            for i, cell in enumerate(self.board)
            if cell == " "
        )

    def play(self, position):
        board = list(self.board)
        board[position] = self.player

        next_player = (
            "O"
            if self.player == "X"
            else "X"
        )

        return GameState(
            board=tuple(board),
            player=next_player
        )
```

```python
WINNING_LINES = (
    (0, 1, 2),
    (3, 4, 5),
    (6, 7, 8),

    (0, 3, 6),
    (1, 4, 7),
    (2, 5, 8),

    (0, 4, 8),
    (2, 4, 6),
)


def winner(state):
    for a, b, c in WINNING_LINES:

        if (
            state.board[a] != " "
            and state.board[a]
            == state.board[b]
            == state.board[c]
        ):
            return state.board[a]

    return None
```

```python
def is_draw(state):
    return (
        winner(state) is None
        and not state.possible_moves()
    )
```

```python
def is_terminal(state):
    return (
        winner(state) is not None
        or is_draw(state)
    )
```

```python
# "Quão boa é essa posição para o jogador MAX?"
def evaluate(state):

    result = winner(state)

    if result == "X":
        return 1

    if result == "O":
        return -1

    return 0
```

```python
def minimax(state):

    if is_terminal(state):
        return evaluate(state)

    if state.player == "X":

        return max(
            minimax(state.play(move))
            for move in state.possible_moves()
        )

    else:

        return min(
            minimax(state.play(move))
            for move in state.possible_moves()
        )
```

```python
def best_move(state):

    moves = state.possible_moves()

    return max(
        moves,
        key=lambda move:
            minimax(state.play(move))
    )
```

```python
state = GameState(
    board=(
        "X", "O", " ",
        " ", "X", " ",
        "O", " ", " "
    ),
    player="X"
)

move = best_move(state)

print("Melhor jogada:", move)
```
