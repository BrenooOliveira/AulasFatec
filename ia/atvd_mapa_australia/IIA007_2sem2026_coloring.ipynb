```python
from simpleai.search import CspProblem, backtrack
```

```python
# Definindo a função que impõe a restrição
def constraint_func(names, values):
    return values[0] != values[1]  
```

```python
# Especificando as variáveis 
regioes = ('AO', 'TN', 'Q', 'NGS', 'V', 'AM', 'T')
```

```python
# Definindo as cores possíveis
colors = dict((regiao, ['cinza-claro', 'cinza-medio', 'cinza-escuro']) for regiao in regioes)
```

```python
# Definindo as restrições do mapa da Austrália
constraints = [
    (('AM', 'AO'), constraint_func),
    (('AM', 'TN'), constraint_func),
    (('AM', 'Q'), constraint_func),
    (('AM', 'NGS'), constraint_func),
    (('AM', 'V'), constraint_func),
    (('AO', 'TN'), constraint_func),
    (('TN', 'Q'), constraint_func),
    (('Q', 'NGS'), constraint_func),
    (('NGS', 'V'), constraint_func),
]
```

```python
# Resolvendo o problema
problem = CspProblem(names, colors, constraints)
```

```python
# Apresentando a solução
output = backtrack(problem)
print('\nColor mapping:\n')
for k, v in output.items():
    print(k, '==>', v) 
```
